# Relationships summary

Simple summary of `relationships.json`: relations identified, visualisation pairs, and relation types with their pairs.

```python
import json
from collections import defaultdict

edges = json.load(open("relationships.json"))["edges"]
```

## Relations identified

```python
print("Relations identified:", len(edges))
```

## Visualisation pairs

Unique source–target pairs (a pair can carry more than one relation).

```python
pairs = defaultdict(list)
for e in edges:
    pairs[(e["source"], e["target"])].append(e["name"])

print("Visualisation pairs:", len(pairs))
print("Visualisations involved:", len({v for p in pairs for v in p}))
print()
# for (s, t), rels in pairs.items():
#     print(f"{s} -> {t}   [{', '.join(rels)}]")
```

## Relation names, their visualisation pairs, and numbers

```python
by_name = defaultdict(list)
for e in edges:
    by_name[e["name"]].append((e["source"], e["target"]))

for name, ps in sorted(by_name.items(), key=lambda kv: -len(kv[1])):
    print(f"{name}: {len(ps)}")
    for s, t in ps:
        print(f"    {s} -> {t}")
    print()
```

## Write everything to `relationships_summary.md`

```python
lines = [
    "# Relationships summary",
    "",
    f"- **Relations identified:** {len(edges)}",
    f"- **Visualisation pairs:** {len(pairs)}",
    f"- **Visualisations involved:** {len({v for p in pairs for v in p})}",
    "",
    "## Relation names",
    "",
    "| Relation | Count |",
    "|---|---|",
]
ordered = sorted(by_name.items(), key=lambda kv: -len(kv[1]))
lines += [f"| {name} | {len(ps)} |" for name, ps in ordered]

lines += ["", "## Relation names and their visualisation pairs", ""]
for name, ps in ordered:
    lines += [f"### {name} ({len(ps)})", ""]
    lines += [f"- `{s}` → `{t}`" for s, t in ps]
    lines.append("")

lines += ["## Visualisation pairs", "", "| Source | Target | Relations |", "|---|---|---|"]
lines += [f"| `{s}` | `{t}` | {', '.join(rels)} |" for (s, t), rels in pairs.items()]

with open("relationships_summary.md", "w") as f:
    f.write("\n".join(lines) + "\n")
print("Wrote relationships_summary.md")
```
